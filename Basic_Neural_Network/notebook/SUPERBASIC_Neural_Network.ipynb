```python
import numpy as np
```

```python
X = np.array([
    [1,1],
    [1,2],
    [2,1],
    [2,7],
    [8,8],
    [9,6],
    [8,9],
    [9,7]
])

y = np.array([
    0,
    0,
    0,
    0,
    1,
    1,
    1,
    1
])
```

```python
weights = np.array([.5, -0.2])
bias = -1
```

```python
z=np.dot(X, weights) + bias
```

```python
print(z)
```

```python
def sigmoid(x):
    return 1 / (1+ np.exp(-x))
```

```python
output = sigmoid(z)
print(output)
```

```python
predictions = (output>=0.5).astype(int)
print(predictions)
```
